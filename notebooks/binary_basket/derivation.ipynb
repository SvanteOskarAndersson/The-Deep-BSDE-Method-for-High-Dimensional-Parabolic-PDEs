Ito formula for $dS_t = \mu\,dt + \sigma\,dW_t$:
$$
du(t,S_t) = \Big[\partial_t u + \mu\cdot\nabla u + \tfrac12\operatorname{Tr}(\sigma\sigma^\top\nabla^2 u)\Big]dt + (\nabla u)^\top\sigma\,dW_t
$$

PDE: $\;\partial_t u + \tfrac12\operatorname{Tr}(\sigma\sigma^\top\nabla^2 u) + \mu\cdot\nabla u - ru = 0$:


Comparing terms gives the corresponding SDE $\;dS_t = \operatorname{diag}(r-q)S_t\,dt + \operatorname{diag}(S_t)\operatorname{diag}(\sigma)\,dW_t$

If we define: $\;Y_t = u(t,S_t), \quad Z_t = \sigma^\top\nabla u, \quad Z^i_t = \sigma_i S^i_t\,\partial_i u$

Itos formula for this PDE becomes:
$$
dY_t = rY_t\,dt + Z_t^\top dW_t, \qquad Y_T = g(S_T), \qquad Y_0 = u(0,\xi)
$$

Discretized version ($\Delta W_n \sim N(0,\Delta t\, I_d)$):
$$
Y_{n+1} = (1+r\Delta t)\,Y_n + Z_n^\top \Delta W_n
$$

$Y_0, Z_0$ trainable, $Z_n = \mathrm{NN}_n(S_n)$, loss $= \frac1M\sum_m \big(Y_N^{(m)} - g(S_N^{(m)})\big)^2$.
